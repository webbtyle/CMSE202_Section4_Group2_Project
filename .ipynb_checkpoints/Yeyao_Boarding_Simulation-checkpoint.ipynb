```python
import random
import time
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Patch
from IPython.display import clear_output
```

```python
# Based on Boeing 777 v3 layout (rows 1–47)
# Reference: https://www.seatlink.com/airlines/united-airlines/seat-maps/united-airlines-boeing-777-v3-seat-map/
# Our final structure:
# Row 1–4   First (2 seat at the left, 4 seats at meddle, 2 seats at the right)
# Row 5–6   Lav/Galley (unavailable)
# Row 7–15  Business cabin (3 seats at the left, 4 seats at meddle, 3 seats at the right)
# Row 16–17 Lav/Galley (unavailable)
# Row 18–30 Premium cabin (3 seats at the left, 4 seats at meddle, 3 seats at the right)
# Row 31–32 Lav/Galley (unavailable)
# Row 33–45 Economy cabin (3 seats at the left, 4 seats at meddle, 3 seats at the right)

# token:
# seat letter = usable seat
# "_" = aisle
# "." = unusable seat/space

# 12 columns per row:
# 0:column A  1:column B  2:column C-or-dot  3:_  4:column D  5:column E  6:colulmn F  7:column G  8:_  9:column J-or-dot  10: column K  11:column L

# Seat Arrangement
NUM_COLS = 12
REAL_777_PATTERN = {}
# Row 1–4
# First (2 seat at the left, 4 seats at meddle, 2 seats at the right): A B . _ D E F G _ . K L
for r in range(1, 5):
    REAL_777_PATTERN[r]=["A","B",".","_","D","E","F","G","_",".","K","L"] # "_" = aisle and "." = unusable seat/space

# Row 5–6
# Lav/Galley (unavailable)
for r in range(5, 7):
    REAL_777_PATTERN[r]=[".",".",".",".",".",".",".",".",".",".",".","."] # "." = unusable seat/space

# Row 7–15
# Business cabin (3 seats at the left, 4 seats at meddle, 3 seats at the right)
for r in range(7, 16):
    REAL_777_PATTERN[r]=["A","B","C","_","D","E","F","G","_","J","K","L"] # "_" = aisle

# Row 16–28
# Premium cabin (3 seats at the left, 4 seats at meddle, 3 seats at the right)
for r in range(16, 29):
    REAL_777_PATTERN[r]=["A","B","C","_","D","E","F","G","_","J","K","L"] # "_" = aisle

# Row 29–30 Lav/Galley (unavailable)
for r in range(29, 31):
    REAL_777_PATTERN[r]=[".",".",".",".",".",".",".",".",".",".",".","."]# "." = unusable seat/space

# Row 31–45 
# Economy cabin (3 seats at the left, 4 seats at meddle, 3 seats at the right)
for r in range(31, 46):
    REAL_777_PATTERN[r]=["A","B","C","_","D","E","F","G","_","J","K","L"]# "_" = aisle
```

```python
# Passenger
# Change 
class Passenger:
    def __init__(self, pid, row, seat, cabin_class,
                 has_carryon,
                 is_senior=False, is_parent=False, is_disabled=False):
        self.pid = pid # Unique passenger ID number (e.g., 1, 2, 3...)
        self.row = row # The target row where the passenger will sit (Row 1–45 on a Boeing 777)
        self.seat = seat # The seat column name assigned to the passenger (A, B, C, D, E, F, G, J, K, or L)

        # Cabin class used for boarding priority ordering:
        #   "First": rows 1–4
        #   "Business": rows 7–15
        #   "Premium": rows 16–28
        #   "Economy": rows 31–45
        self.cabin_class = cabin_class 
        
        self.has_carryon = has_carryon # Whether this passenger is bringing a carry-on bag. Only passengers with carry-on will experience luggage stowing delay.

        # Add other priority conditions
        self.is_senior = is_senior # Elderly passenger
        self.is_parent = is_parent # Adult traveling with children
        self.is_disabled = is_disabled # Passenger with disabilities
        self.is_epd = is_senior or is_parent or is_disabled  # Elder, Parents, Disabled

        # luggage handling
        self.luggage_delay = random.randint(9, 12) # Set number of steps needed to place luggage in the overhead bin. At least 9 seconds and at most 12 seconds to place 1 luggage
        self.remaining_delay = self.luggage_delay # Remaining luggage time in seconds (counts down as simulation runs)
        self.carryon_stowed = (not has_carryon) # If a passenger has no carry-on, will be considered "stowed"

        # Set dynamic position
        self.x = -1   # Current x-position in the aisle, default x=-1 to indicate that a passenger has not entered yet
        self.y = None # Current y-position (column index), which will be set when the passenger enters the plane.
        self.seated = False # Will be used to define whether a passenger is already seated (either True or False)
        
    # We want to show a passenger is stowing luggage
    def is_putting_luggage(self):
        # If the passenger has no stowing left to do OR already completed stowing,
        # they cannot be considered as "putting luggage".
        if self.carryon_stowed or self.remaining_delay <= 0:
            return False
        # Only considered "putting luggage" if they are exactly at their target row.
        return self.x == self.row - 1 # E.g. a passenger assigned to row 18 should be at aisle coordinate x = 17.

    def __repr__(self):
        return f"Passenger({self.row}{self.seat}, {self.cabin_class}, seated={self.seated})"
```

```python
# PlaneLayout
class PlaneLayout:
    def __init__(self, row_patterns):
        self.row_pattern = row_patterns # later will be applied to REAL_777_PATTERN
        self.total_rows = max(row_patterns.keys()) # find the largest row number, which will be 45 rows in REAL_777_PATTERN
        #AI assistance used to understand how to construct structured data mappings for a
    # Boeing 777 seat layout. The goal was to learn how to translate raw seat-pattern
    # tokens (e.g., ['A','B','C','_',...]) into computational structures such as:
    # (1) seat-letter-to-column maps (y_map),
    # (2) aisle-column indices (aisles),
    # (3) categorical seat types (seat_type),
    # (4) per-row valid seating lists (row_seats).
    # Source: OpenAI. (2025). ChatGPT-5. Retrieved from https://chatgpt.com/, Nov 20, 2025.
    # Prompt: "Provide a full walkthrough of the PlaneLayout class and explain the role
    # of y_map, aisles, seat_type, and row_seats, including how they are computed and
    # why they are needed for a plane boarding simulation."

        self.y_map = {}      # Maps each row to {seat_letter will be column_index} (A,B,C,D,E,F,G,H,I,J,K, and L)
        self.aisles = {}     # Maps each row to a list of aisle column indices
        self.seat_type = {}  # Maps each seat to its type: window, middle, or aisle
        self.row_seats = {}  # Lists all valid seat letters in each row 
        self._build()        # Parse row_patterns to fill the above structures 

        # aisles from row 1 (two aisles)
        # Use row 1 to detect aisle positions, where column index 3 and 8 will be aisle position
        # Set the left aisle as the passenger boarding entrance (y=3)
        self.all_aisles = self.aisles[1]  # Aisle locations are the same for all rows, so take them from row 1
        self.entrance_y = self.all_aisles[0] # default: will be enter from left aisle

    def _build(self):
        # Loop through each row in the plane and parse its seat/aisle pattern
        for row, pattern in self.row_pattern.items():
            y_mapping = {} # {"A":0, "B":1, "C":2, "D":4 …}
            ### Column 3 and Column 8 are aisles
            aisles = [] # Maps each row to a list of aisle column indices, which aisles will be [3,8] in REAL_777_PATTERN later
            seat_type = {} # Maps each seat to its type: window, middle, or aisle

            # Scan each column in the row pattern
            for y, token in enumerate(pattern):
                # If a column in a row is an aisle, will record its location in aisles
                if token == "_":
                    aisles.append(y) 
                # If a column in a row is an unusable space (lavatory/galley), will skip
                elif token == ".":
                    continue
                # The rest will be real seat letter, we will record its y-coordinate    
                else:
                    y_mapping[token] = y

            # classify window or middle seat
            if y_mapping:
                # Identify leftmost and rightmost seats will be identified as window seats
                min_y = min(y_mapping.values()) 
                max_y = max(y_mapping.values())
                
                for seat, yy in y_mapping.items():
                    if yy == min_y or yy == max_y:
                        seat_type[seat] = "window" # e.g., A and L
                    else:
                        seat_type[seat] = "middle" # default until aisle check

                # Identify aisle-adjacent seats, which will be true aisle seats
                for ay in aisles:
                    for seat, yy in y_mapping.items():
                        if abs(yy - ay) == 1:
                            seat_type[seat] = "aisle"
                            
            # Store computed structures for this row
            self.y_map[row] = y_mapping # seat letter to be y coordinate
            self.aisles[row] = aisles # aisle column indices for this row
            self.seat_type[row] = seat_type # seat type classification
            self.row_seats[row] = list(y_mapping.keys()) # all valid seat letters in this row
            
    # mapping functions 
    def seat_y(self, row, seat):
        return self.y_map[row][seat]

    def cabin_class_of_row(self, row):
        # Use cabin_class set boarding priority
        if 1 <= row <= 4:
            return "first"
        elif 7 <= row <= 15:
            return "second"
        elif 18 <= row <= 30:
            return "third"
        elif 33 <= row <= 47:
            return "fourth"
        else:
            return None  # lavatory/galley rows

    def cabin_region(self, row):
        # set background color
        if 1 <= row <= 4:
            return "first"
        elif 7 <= row <= 15:
            return "second"
        elif 18 <= row <= 30:
            return "third"
        elif 33 <= row <= 47:
            return "fourth"
        else:
            return "lav"

    def max_y(self):
        return NUM_COLS - 1
```

```python
class BoardingStrategy:
    """
    EXACT order:
    For each cabin (first → second → third → fourth):
        1) EPD:
            rows from BACK → FRONT
                window → middle → aisle
        2) non-EPD:
            rows from BACK → FRONT
                window → middle → aisle
    """

    def sort_passengers(self, passengers, layout):

        cabin_order = ["first", "second", "third", "fourth"]
        seat_order = ["window", "middle", "aisle"]

        final_order = []

        for cabin in cabin_order:

            # For EPD passengers
            epd_group = [p for p in passengers if p.cabin_class == cabin and p.is_epd]
            if epd_group:
                rows = sorted({p.row for p in epd_group}, reverse=True)
                for row in rows:
                    for seat_type in seat_order:
                        for p in epd_group:
                            if p.row == row and layout.seat_type[p.row][p.seat] == seat_type:
                                final_order.append(p)

            # For Non-EPD passengers
            non_epd_group = [p for p in passengers if p.cabin_class == cabin and not p.is_epd]
            if non_epd_group:
                rows = sorted({p.row for p in non_epd_group}, reverse=True)
                for row in rows:
                    for seat_type in seat_order:
                        for p in non_epd_group:
                            if p.row == row and layout.seat_type[p.row][p.seat] == seat_type:
                                final_order.append(p)

        return final_order
```

```python
# Boarding Simulation
class BoardingSimulation:
    def __init__(self, layout: PlaneLayout, passengers,
                 visualize=False, delay=0.05):
        self.layout = layout
        self.passengers = passengers
        self.waiting_queue = passengers.copy()
        self.time = 0
        self.visualize = visualize
        self.delay = delay

    def cell_occupied(self, x, y):
        return any((not p.seated) and p.x == x and p.y == y
                   for p in self.passengers)

    def try_spawn(self):
        if not self.waiting_queue:
            return

        p = self.waiting_queue[0]
        target_y = self.layout.seat_y(p.row, p.seat)
        aisles = self.layout.all_aisles
        entry_y = min(aisles, key=lambda a: abs(a - target_y))

        if not self.cell_occupied(0, entry_y):
            self.waiting_queue.pop(0)
            p.x = 0
            p.y = entry_y

    def step_passenger(self, p: Passenger):
        if p.seated or p.x < 0:
            return

        target_x = p.row - 1

        # Move down aisle
        if p.x < target_x:
            next_x = p.x + 1
            if not self.cell_occupied(next_x, p.y):
                p.x = next_x
            return

        # At own row: handle luggage first
        if not p.carryon_stowed:
            if p.remaining_delay > 0:
                p.remaining_delay -= 1   # red while this is happening
                return
            else:
                p.carryon_stowed = True

        # Move horizontally to seat
        seat_y = self.layout.seat_y(p.row, p.seat)
        if p.y != seat_y:
            step = 1 if seat_y > p.y else -1
            next_y = p.y + step
            if not self.cell_occupied(p.x, next_y):
                p.y = next_y
            return

        # Sit down
        p.seated = True

    def step(self):
        self.try_spawn()
        movers = [p for p in self.passengers if not p.seated and p.x >= 0]
        movers.sort(key=lambda p: (p.x, p.row), reverse=True)
        for p in movers:
            self.step_passenger(p)
        self.time += 1

    def run(self, visualizer=None):
        step_count = 0
        while not all(p.seated for p in self.passengers):
            if self.visualize and visualizer is not None:
                clear_output(wait=True)
                visualizer.show(step_count)
                time.sleep(self.delay)
            self.step()
            step_count += 1

        if self.visualize and visualizer is not None:
            clear_output(wait=True)
            visualizer.show(step_count)

        print(f"All passengers seated in {self.time} time steps.")
        return self.time
```

```python
# Visualizer
UNAVAILABLE_COLOR = "#D9D9D9"
AISLE_COLOR = "#D9D9D9"

class BoardingVisualizer:
    def __init__(self, layout: PlaneLayout, passengers):
        self.layout = layout
        self.passengers = passengers

    # base color by cabin & EPD
    def base_color(self, p: Passenger):
        if p.cabin_class == "first":
            return "green" if p.is_epd else "yellow"
        elif p.cabin_class == "second":  # premium
            return "pink" if p.is_epd else "blue"
        elif p.cabin_class in ("third", "fourth"):
            return "lightskyblue" if p.is_epd else "black"
        return "grey"

    def draw_cabin_backgrounds(self, ax):
        # (start_row, end_row, color, label)
        regions = [
            (1, 4,  "#FFF7C2", "First zone"),
            (7, 15, "#D6E9FF", "Second zone"),
            (18, 30,"#FFFFFF", "Third zone"),
            (33, 47,"#E8FCD7", "Fourth zone"),
        ]
        for (start, end, color, _) in regions:
            ax.axvspan(start - 1, end, facecolor=color, alpha=0.3, zorder=0)

    def show(self, step_num=None):
        fig, ax = plt.subplots(figsize=(self.layout.total_rows/2.5, 5))
        seat_size = 0.85

        # Set color to different class zones
        self.draw_cabin_backgrounds(ax)

        # grids included aisles and unavailable seats
        for row in range(1, self.layout.total_rows + 1):
            pattern = self.layout.row_pattern[row]
            for y, token in enumerate(pattern):
                if token == "_":
                    face = AISLE_COLOR
                elif token == ".":
                    face = UNAVAILABLE_COLOR
                else:
                    face = "white"
                ax.add_patch(Rectangle((row - 1, y), seat_size, seat_size,
                                       facecolor=face, edgecolor="black",
                                       zorder=1))

        # passengers who are sitting on their seats
        for p in self.passengers:
            if p.seated:
                color = self.base_color(p)
                y = self.layout.seat_y(p.row, p.seat)
                ax.add_patch(Rectangle((p.row - 1, y), seat_size, seat_size,
                                       facecolor=color, edgecolor="black",
                                       zorder=3))

        # passengers who are walking or placing their bags
        for p in self.passengers:
            if (not p.seated) and p.x >= 0:
                color = "red" if p.is_putting_luggage() else self.base_color(p)
                ax.scatter(p.x + seat_size/2, p.y + seat_size/2,
                           c=color, s=80, edgecolors="black", zorder=4)

        # set x and y axis, where x=column, y= rows
        ax.set_xlim(-1, self.layout.total_rows)
        ax.set_ylim(NUM_COLS, -1)   # invert Y so rows started from A appears
        ax.invert_yaxis()

        ax.set_xticks(range(self.layout.total_rows))
        ax.set_xticklabels(range(1, self.layout.total_rows + 1), rotation=90)

        y_ticks = list(range(NUM_COLS))
        y_labels = ["A","B","C","","D","E","F","G","","J","K","L"]
        ax.set_yticks(y_ticks)
        ax.set_yticklabels(y_labels)

        for spine in ["top", "right"]:
            ax.spines[spine].set_visible(False)

        title = "Plane Boarding"
        if step_num is not None:
            title += f" - Step {step_num}"
        ax.set_title(title)

        # Legend
        legend_handles = [
            Patch(facecolor="yellow", edgecolor="black", label="First normal"),
            Patch(facecolor="green",  edgecolor="black", label="First EPD"),
            Patch(facecolor="blue",   edgecolor="black", label="Second normal"),
            Patch(facecolor="pink",   edgecolor="black", label="Second EPD"),
            Patch(facecolor="black",  edgecolor="black", label="Economy normal"),
            Patch(facecolor="lightskyblue", edgecolor="black", label="Economy EPD"),
            Patch(facecolor="red", edgecolor="black", label="Luggage stowing"),
            Patch(facecolor=AISLE_COLOR, edgecolor="black", label="Aisle"),
        ]
        ax.legend(handles=legend_handles, bbox_to_anchor=(1.02, 1),
                  loc="upper left", borderaxespad=0.)
        plt.tight_layout()
        plt.show()
```

```python
# Passenger generator
# set possibilities for each condition happens on a passenger
def create_passengers(layout: PlaneLayout,
                      p_carryon=0.7,
                      p_senior=0.05,
                      p_parent=0.1,
                      p_disabled=0.03,
                      seed=0):
    random.seed(seed)
    passengers = []
    pid = 1
    for row in range(1, layout.total_rows + 1):
        cabin = layout.cabin_class_of_row(row)
        if cabin is None:
            continue
        for seat in layout.row_seats[row]:
            p = Passenger(
                pid=pid,
                row=row,
                seat=seat,
                cabin_class=cabin,
                has_carryon=(random.random() < p_carryon),
                is_senior=(random.random() < p_senior),
                is_parent=(random.random() < p_parent),
                is_disabled=(random.random() < p_disabled),
            )
            passengers.append(p)
            pid += 1
    return passengers
```

```python
# Run example
layout = PlaneLayout(REAL_777_PATTERN)
passengers = create_passengers(layout, seed=1007) # feel free to set seeds here

strategy = BoardingStrategy()
sorted_passengers = strategy.sort_passengers(passengers, layout)

visualizer = BoardingVisualizer(layout, sorted_passengers)
sim = BoardingSimulation(layout, sorted_passengers,visualize=True, delay=0.05)

sim.run(visualizer=visualizer)
```

## Southwest Method

### Reference: https://www.vox.com/2014/4/25/5647696/the-way-we-board-airplanes-makes-absolutely-no-sense 

```python
class SouthwestStrategy:
    """Southwest Airlines: boarding order is completely random."""
    def sort_passengers(self, passengers, layout):
        arr = passengers.copy()
        random.shuffle(arr)
        return arr
```

```python
layout = PlaneLayout(REAL_777_PATTERN)
passengers = create_passengers(layout, seed=1007)

strategy = SouthwestStrategy()
ordered = strategy.sort_passengers(passengers, layout)

visualizer = BoardingVisualizer(layout, ordered)
sim = BoardingSimulation(layout, ordered, visualize=True, delay=0.05)

sim.run(visualizer)
```

# Steffen Method

### Reference: https://www.vox.com/2014/4/25/5647696/the-way-we-board-airplanes-makes-absolutely-no-sense

```python
class SteffenStrategy:
    """Steffen Method adapted to the actual 777 seat map."""
    def sort_passengers(self, passengers, layout):

        # Group by seat type, not seat letter
        window = []
        middle = []
        aisle  = []

        for p in passengers:
            stype = layout.seat_type[p.row][p.seat]
            if stype == "window":
                window.append(p)
            elif stype == "middle":
                middle.append(p)
            elif stype == "aisle":
                aisle.append(p)

        # Standard Steffen staggered order function
        def steffen_order(group):
            rows = sorted({p.row for p in group}, reverse=True)
            even = [r for r in rows if r % 2 == 0]
            odd  = [r for r in rows if r % 2 == 1]
            ordered_rows = even + odd
            result = []
            for r in ordered_rows:
                result.extend([p for p in group if p.row == r])
            return result

        final = []
        final += steffen_order(window)
        final += steffen_order(middle)
        final += steffen_order(aisle)

        return final
```

```python
layout = PlaneLayout(REAL_777_PATTERN)
passengers = create_passengers(layout, seed=1007)

strategy = SteffenStrategy()
ordered = strategy.sort_passengers(passengers, layout)

visualizer = BoardingVisualizer(layout, ordered)
sim = BoardingSimulation(layout, ordered, visualize=True, delay=0.05)

sim.run(visualizer)
```

## Calculate the average steps for each strategy

```python
def run_multiple_times_random_seed(layout, strategy_class, runs=1000):
    """Run boarding simulation many times with random seeds and return average steps."""
    times = []

    for _ in range(runs):
        seed_value = random.randint(1, 10**9)  # generate a true random seed
        passengers = create_passengers(layout, seed=seed_value)

        strategy = strategy_class()
        ordered = strategy.sort_passengers(passengers, layout)

        sim = BoardingSimulation(layout, ordered, visualize=False)
        t = sim.run()
        times.append(t)

    avg = sum(times) / len(times)
    return avg, times
```

```python
# Strategy setted by us
layout = PlaneLayout(REAL_777_PATTERN)
avg, steps_list = run_multiple_times_random_seed(layout,BoardingStrategy , runs=100)
print(f"Setted borading strategy average steps: {avg:.2f}")
```

```python
# Southwest
layout = PlaneLayout(REAL_777_PATTERN)
avg, steps_list = run_multiple_times_random_seed(layout,SouthwestStrategy , runs=100)
print(f"SouthWest Strategy average steps: {avg:.2f}")
```

```python
# Steffen
layout = PlaneLayout(REAL_777_PATTERN)
avg, steps_list = run_multiple_times_random_seed(layout, SteffenStrategy, runs=100)
print(f"Steffen average steps: {avg:.2f}")
```
