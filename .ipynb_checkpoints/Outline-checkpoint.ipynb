## Outline: Plane Boarding

**Scenario Description**

* 20 Rows
* 2 rows on each side of the plane with a central walkway
* The first 5 rows will be first/business class
* Each seat is one "space" and the aisle is also one space for a width of 5 spaces for each row
* They will load at the front left of the plane
* Passengers can load luggage in the overhead, but it takes a certain number of steps
* A random number of passengers will have carryons to put in the overhead
* Passengers can move forward, left, right, but not backward. They can look forward and left and right. They can put carryon in the overhead.
* Passengers attributes will include their seat location/number, class of seating, whether they have a carryon or not, senior status, parent status, disability status

## Boarding Methods

1. People with carryons board last
2. People with carryons board first
3. Elderly, parents, disabled board first/last
4. Window seats board first followed by the aisle
5. Board in seat order back to front
6. Board in seat order front to back

Increase steps for loading carryons for people with kids and elderly
