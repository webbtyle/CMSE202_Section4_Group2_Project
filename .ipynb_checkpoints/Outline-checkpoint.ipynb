## Outline: Plane Boarding

**Scenario Description**

* 20 Rows
* Plane is 6 seats wide, 3–aisle–3 configuration. 3 closer to aisle on either side, aisle in centre. 6 seats total per row, 1 space for each seat, plus aisle for a total width of 5 spaces per row
* First five rows will be first/business class
* Will test several methods which all arrange different priorities for passengers to seat
* Passengers can load luggage in overhead, but it takes a certain number of steps, and for sake of argument we’ll say 9-12 seconds loading each luggage
* A random number of passengres will have carryons to put in overhead
* Passengers can move forward, left, right, but not backwards, can look forward, and left and right, and can put carryon in overhead
* Passengers attributes: seat location/number, class of seating, whether have carryon or not, senior status, parent status, disability status (Optional)

## Boarding Methods

1. Board in seat order front to back
2. Board in seat order back to front
3. WILMA Method
4. Southwest Method
5. Steffen Method
